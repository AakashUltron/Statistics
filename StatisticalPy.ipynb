### The length of a human pregnancy is normally distributed with a mean of 272 days with a Standard Deviation of 9 days.

**1. State the Random Variable.**

**2. Find the probablity of a pregnancy lasting more than 280 days.**

**3. Find the probablity of a pregnancy lasting more than 250 days.**
 
**4. Find the probablity of a pregnancy lasts between 265 and 280 days.**

### Solution

**1. RV - Length of Human Pregnancy.**

```python
from scipy import stats
```

**2. Find the probablity of a pregnancy lasting more than 280 days.**

```python
1 - stats.norm.cdf(x = 280, loc = 272, scale = 9)
```

**3. Find the probablity of a pregnancy lasting more than 250 days.**

```python
stats.norm.cdf(x = 250, loc = 272, scale = 9)
```

**4. Find the probablity of a pregnancy lasts between 265 and 280 days.**

```python
stats.norm.cdf(x = 280, loc = 272, scale = 9) - stats.norm.cdf(x = 265, loc = 272, scale = 9)
```

### A company claims that the average time taken to resolve customer tickets is 30 minutes. A random sample of 40 tickets shows a mean resolution time of 33 minutes with a standard deviation of 8 minutes. At the 5% level of significance, test whether the average resolution time has increased.

- H0 - Current Process/Statement of No change - There is no significant difference in the average resolution time.
- H1 - Statement of change/New Process - There is a singnificant difference in the average resolution time

```python
from scipy import stats
import numpy as np
```

```python
observations_40_tickets = np.random.normal(loc = 33, scale = 5, size = 40).astype("int")
observations_40_tickets
```

```python
t_stats,p_value = stats.ttest_1samp(a= observations_40_tickets, popmean = 30, alternative = 'greater')
t_stats,p_value
```

```python
p_value = p_value/2  # Note - for one tailed test always p_value should be divide by 2.
```

```python
if p_value < 0.05:
    print("We Reject H0")
else:
    print("We consider H1")
```

#### At 5% level of significance, we can reject the Null Hypothesis and claim that average resolution time has increased

7.	Two versions of a website are tested. Out of 500 users for Version A, 320 users click the signup button. Out of 480 users for Version B, 290 users click the signup button. At the 5% level of significance, test whether the conversion rates differ.

**UNIVERSAL STAT TEST TEMPLATE**

```python
import math
from scipy.stats import t, norm

# ---------------------------
# 🔹 SELECT TEST TYPE
# ---------------------------
test_type = "t_two_sample"  
# Options:
# "t_one_sample"
# "t_two_sample"
# "z_one_proportion"
# "z_two_proportion"

alpha = 0.05   # significance level

# ---------------------------
# 🔹 INPUT VALUES
# ---------------------------

# For t-tests
mean1, std1, n1 = 48, 6, 35
mean2, std2, n2 = 52, 7, 40
mu0 = 65   # for one-sample t-test

# For proportions
x1, n1_p = 320, 500
x2, n2_p = 290, 480
p0 = 0.60  # for one-sample proportion

# ---------------------------
# 🔹 CALCULATION
# ---------------------------

if test_type == "t_one_sample":
    t_stat = (mean1 - mu0) / (std1 / math.sqrt(n1))
    df = n1 - 1
    p_value = 2 * (1 - t.cdf(abs(t_stat), df))

elif test_type == "t_two_sample":
    t_stat = (mean1 - mean2) / math.sqrt((std1**2/n1) + (std2**2/n2))
    df = n1 + n2 - 2
    p_value = 2 * (1 - t.cdf(abs(t_stat), df))

elif test_type == "z_one_proportion":
    p_hat = x1 / n1_p
    z_stat = (p_hat - p0) / math.sqrt((p0*(1-p0))/n1_p)
    p_value = 2 * (1 - norm.cdf(abs(z_stat)))

elif test_type == "z_two_proportion":
    p1 = x1 / n1_p
    p2 = x2 / n2_p
    p = (x1 + x2) / (n1_p + n2_p)
    z_stat = (p1 - p2) / math.sqrt(p*(1-p)*(1/n1_p + 1/n2_p))
    p_value = 2 * (1 - norm.cdf(abs(z_stat)))

# ---------------------------
# 🔹 OUTPUT
# ---------------------------

try:
    print("t-statistic:", t_stat)
except:
    print("z-statistic:", z_stat)

print("p-value:", p_value)

# ---------------------------
# 🔹 DECISION
# ---------------------------

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```
