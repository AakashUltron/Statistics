```python
import numpy as np
from scipy import stats
```

## 1. The historical average score in a certification exam is 65. A training institute introduces a new course. A random sample of 50 students trained under the new course has a mean score of 68 and a standard deviation of 10. At the 5% level of significance, test whether the new course improves  performance.

```python
sample = np.random.normal(loc = 65, scale = 10, size = 50).astype("int")
sample
```

```python
t_stat, p_value_two_tailed = stats.ttest_1samp(a = sample, popmean = 65)
t_stat, p_value_two_tailed
```

```python
p_value_one_tailed = p_value_two_tailed / 2
p_value_one_tailed
```

```python
print("t-statistic:", t_stat)
print("p-value (one-tailed):", p_value_one_tailed)
```

```python
alpha = 0.05
if (t_stat > 0) and (p_value_one_tailed < alpha):
    print("Reject H0 → Course improves performance")
else:
    print("Fail to reject H0")
```

### 2. A company claims that the average time taken to resolve customer tickets is 30 minutes. A random sample of 40 tickets shows a mean resolution time of 33 minutes with a standard deviation of 8 minutes. At the 5% level of significance, test whether the average resolution time has increased.

```python
p_value = p_value/2 
```

```python
if p_value < 0.05:
    print("We Reject H0")
else:
    print("We consider H1")
```

## 3. A company claims that 60% of its users are satisfied with its product. In a survey of 200 users, 134 users report satisfaction. At the 5% level of significance, test whether customer satisfaction is higher than claimed.

```python
from statsmodels.stats.proportion import proportions_ztest
```

```python
z_stat, p_value_two_tailed = proportions_ztest(count = 134, nobs = 200, value= 0.60)
z_stat, p_value_two_tailed
```

```python
p_value_one_tailed = p_value_two_tailed / 2
p_value_one_tailed
```

```python
print("z-statistic:", z_stat)
print("p-value (one-tailed):", p_value_one_tailed)
```

```python
alpha = 0.05
if (z_stat > 0) and (p_value_one_tailed < alpha):
    print("Reject H0 → Satisfaction is higher than claimed")
else:
    print("Fail to reject H0")
```

## 5. Two teaching methods are used to train employees. A sample of 30 employees trained using Method A has a mean score of 75 with a standard deviation of 12. A sample of 28 employees trained using Method B has a mean score of 70 with a standard deviation of 10.
```python
print("t-statistic:", t_stat)
print("p-value:", p_value_two_tailed)
```

```python
alpha = 0.05
if p_value_two_tailed < alpha:
    print("Reject H0 → There is a difference")
else:
    print("Fail to reject H0")
```

### 7. Two versions of a website are tested. Out of 500 users for Version A, 320 users click the signup button. Out of 480 users for Version B, 290 users click the signup button. At the 5% level of significance, test whether the conversion rates differ.

```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest

# Given data
count = np.array([320, 290])
nobs = np.array([500, 480])

# Two-proportion z-test
z_stat, p_value_two_tailed = proportions_ztest(count, nobs)

print("z-statistic:", z_stat)
print("p-value:", p_value_two_tailed)

# Decision
alpha = 0.05
if p_value_two_tailed < alpha:
    print("Reject H0 → Conversion rates differ")
else:
    print("Fail to reject H0")
```

### 9.	A company wants to compare the average productivity of employees trained using three different training programs. Program A has 25 employees with a mean productivity score of 80 and a standard deviation of 8. Program B has 30 employees with a mean of 75 and a standard deviation of 7. Program C has 28 employees with a mean of 82 and a standard deviation of 6. At the 5% level of significance, test whether there is a difference in mean productivity across the programs.

```python
np.random.seed(0)

A = np.random.normal(loc = 80, scale = 8, size = 25).astype("int")
B = np.random.normal(loc = 75, scale = 7, size = 30).astype("int")
C = np.random.normal(loc = 82, scale = 6, size = 28).astype("int")
```

```python
print(A)
print(B)
print(C)
```

```python
f_stat, p_value = stats.f_oneway(A,B,C)
f_stat, p_value
```

```python
print("F-statistic:", f_stat)
print("p-value:", p_value)
```

```python
alpha = 0.05
if p_value < alpha:
    print("Reject H0 → Productivity differs")
else:
    print("Fail to reject H0")
```

### 10.	A company wants to test whether gender and product preference are independent. The survey results are as follows:
-	Male: 60 prefer Product A, 40 prefer Product B
-	Female: 50 prefer Product A, 50 prefer Product B

**At the 5% level of significance, test for independence.**


```python
import numpy as np
from scipy.stats import chi2_contingency

observed = np.array([[60, 40],
                     [50, 50]])

chi_stat, p_value, dof, expected = chi2_contingency(observed)

print("Chi-square:", chi_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject H0 → Dependent")
else:
    print("Fail to reject H0 → Independent")
```

### 12.	A dice manufacturer claims that a die is fair. The die is rolled 120 times, and the observed frequencies are:
-	Face 1: 18
-	Face 2: 22
-	Face 3: 20
-	Face 4: 25
-	Face 5: 17
-	Face 6: 18

**At the 5% level of significance, test whether the die is fair.**


```python
import numpy as np
from scipy.stats import chisquare
```

```python
observation = np.array([18,22,20,25,17,18]) 
observation
```

```python
expectation = np.full(6, observation.sum() / 6)
expectation
```

```python
chi_stat, p_value = chisquare(observation,expectation)
chi_stat, p_value
```

```python
print("Chi-square:", chi_stat)
print("p-value:", p_value)
```

```python
if p_value < 0.05:
    print("Reject H0 → Not fair")
else:
    print("Fail to reject H0 → Fair die")
```
